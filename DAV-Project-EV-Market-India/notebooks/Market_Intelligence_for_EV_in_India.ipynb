### Standardizing Inconsistent Categorical Text (Manufacturer)

```python
# The 'Manufacturer' column has inconsistent casing (e.g., 'mahindra', 'MAHINDRA', 'Mahindra')
print("Unique Manufacturer values BEFORE cleaning:")
print(sorted(df["Manufacturer"].unique()))
```

```python
# Standardize manufacturer names using a lowercase-key mapping so that acronym brands
# (BYD, MG, TVS) are NOT mangled by simple title-casing
manufacturer_map = {
    "byd": "BYD",
    "mahindra": "Mahindra",
    "ather": "Ather",
    "ather energy": "Ather",
    "bajaj": "Bajaj",
    "hyundai": "Hyundai",
    "ola electric": "Ola Electric",
    "ola": "Ola Electric",
    "mg": "MG",
    "tata motors": "Tata Motors",
    "tata": "Tata Motors",
    "tvs": "TVS"
}

# Strip whitespace and lowercase for lookup, then map to the standardized canonical name
df["Manufacturer"] = df["Manufacturer"].str.strip().str.lower().map(manufacturer_map)

print("Unique Manufacturer values AFTER cleaning:")
print(sorted(df["Manufacturer"].unique()))
```

### Observation

- After **standardizing manufacturer names** and merging inconsistent brand variants, the number of unique manufacturers reduced from **18 to 9**, providing a more accurate representation of market competition. Acronym-based brands such as **BYD**, **MG**, and **TVS** were preserved correctly.

### 10.4 Missing Value Treatment

```python
# Numerical columns with missing values -> impute using the MEDIAN (robust to outliers/skew)
numerical_missing_cols = ["Battery_kWh", "Range_km", "Subsidy_INR", "Temperature_C"]

for col in numerical_missing_cols:
    median_value = df[col].median()
    df[col] = df[col].fillna(median_value)
    print(f"{col}: filled {col} missing values with median = {median_value:.2f}")
```

```python
# Categorical column 'Charging_Type' -> impute using the MODE (most frequent category)
charging_mode = df["Charging_Type"].mode()[0]
df["Charging_Type"] = df["Charging_Type"].fillna(charging_mode)
print("Charging_Type: filled missing values with mode =", charging_mode)

# Categorical column 'Festival' -> missing means "no festival period"; fill with a dedicated label
df["Festival"] = df["Festival"].fillna("No Festival")
print("Festival: filled missing values with label = 'No Festival'")
```

```python
# Verify no missing values remain
df.isnull().sum().sort_values(ascending=False)
```

**Observation:** All missing values have been treated — numerical fields used the **median** (robust
to skewed distributions such as price and range), the categorical `Charging_Type` used the **mode**,
and `Festival` was filled with an explicit `"No Festival"` label since its absence is meaningful
rather than erroneous. The dataset now has **zero missing values**.


### Outlier Detection and Treatment

```python
# Visualize potential outliers in key numerical columns using box plots
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

sns.boxplot(y=df["Vehicle_Price_INR"], ax=axes[0], color="skyblue")
axes[0].set_title("Box Plot: Vehicle Price (INR)", fontsize=12)
axes[0].set_ylabel("Vehicle Price (INR)")

sns.boxplot(y=df["Battery_kWh"], ax=axes[1], color="lightgreen")
axes[1].set_title("Box Plot: Battery Capacity (kWh)", fontsize=12)
axes[1].set_ylabel("Battery Capacity (kWh)")

sns.boxplot(y=df["Range_km"], ax=axes[2], color="salmon")
axes[2].set_title("Box Plot: Driving Range (km)", fontsize=12)
axes[2].set_ylabel("Range (km)")

plt.tight_layout()
plt.show()
```

```python
# Dataset size BEFORE outlier removal
print("Dataset shape BEFORE outlier removal:", df.shape)
```

```python
# Apply the IQR (Interquartile Range) method to detect and remove outliers
# from 'Vehicle_Price_INR', the column showing the most extreme outliers
def remove_outliers_iqr(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    return data[(data[column] >= lower_bound) & (data[column] <= upper_bound)]

df = remove_outliers_iqr(df, "Vehicle_Price_INR")
df = remove_outliers_iqr(df, "Battery_kWh")

# Dataset size AFTER outlier removal
print("Dataset shape AFTER outlier removal:", df.shape)
```

```python
# Reset the index after cleaning operations
df.reset_index(drop=True, inplace=True)
df.head()
```

### Observation

- The **IQR method** was used to detect and remove **extreme outliers** in **`Price_INR`** and **`Battery_kWh`**, eliminating unrealistic values (e.g., vehicle prices up to **₹10 crore**) and resulting in a cleaner, more reliable dataset for analysis and visualization.

## **Correlation Heatmap**

```python
# Select numerical columns for correlation analysis
numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()

# Exclude identifier-type columns that carry no analytical meaning
exclude_cols = ["Registration_ID"]
numeric_cols = [c for c in numeric_cols if c not in exclude_cols]

correlation_matrix = df[numeric_cols].corr()

plt.figure(figsize=(16, 12))
sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5,
    square=True,
    cbar_kws={"shrink": 0.8}
)
plt.title("Correlation Heatmap of Numerical Variables", fontsize=15, fontweight="bold")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

**Observations:**
- `Vehicle_Price_INR` and `Range_km` show a **positive correlation**, confirming the pricing pattern
  observed in Q11.
- `Battery_kWh` correlates positively with both `Range_km` and `Vehicle_Price_INR`, as expected from
  underlying EV engineering (bigger batteries → longer range → higher cost).
- Macro-economic indicators such as `Regional_GDP_BnINR` and `Population_Million` show relatively
  **weak correlation** with vehicle-level attributes, since they operate at a different (state) level.
- `Market_Share`, `YoY_Growth`, and `MoM_Growth` show **low mutual correlation**, indicating they
  capture distinct dimensions of market performance.
